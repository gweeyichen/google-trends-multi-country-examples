# Compare one keyword across countries

Runs the Actor in compare mode and charts the interest series, all on one 0-100 scale. Needs `APIFY_TOKEN` in the environment. Running it is billed by Apify.

```python
# pip install apify-client matplotlib
import os
from apify_client import ApifyClient
import matplotlib.pyplot as plt

client = ApifyClient(os.environ["APIFY_TOKEN"])
run = client.actor("asia-data-desk/google-trends-sea").call(run_input={
    "mode": "compare",
    "countryPreset": "SEA",
    "keywords": ["durian"],
    "timeframe": "today 12-m",
    "maxItems": 10,
})
items = [i for i in client.dataset(run.default_dataset_id).iterate_items()
         if i.get("record_type") == "interest" and i.get("has_data")]
len(items)
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
for item in items:
    dates = [p["date"] for p in item["series"]]
    values = [p["value"] for p in item["series"]]
    ax.plot(dates, values, label=item["country"])
ax.set_ylabel("Interest (0-100, one scale)")
ax.set_title("durian, Southeast Asia, last 12 months")
ax.set_xticks(ax.get_xticks()[::8])
ax.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
print("Data source: Google Trends (https://www.google.com/trends)")
```
